```python
with open(result_path + fs_file, "r") as f:
    data = json.load(f)

# Extract layers and after intervention accuracies
layers = list(data.keys())
layers = sorted(layers, key=int)  # Ensure layers are sorted numerically
intervention_accuracies = [data[layer]["intervention_topk"][0][1] for layer in layers]

# Extract baseline (before intervention) accuracies from layer 0
baseline_accuracy = data["0"]["clean_topk"][0][1]
baseline_accuracies = [baseline_accuracy] * len(layers)  # Same value across all layers

# Create Plotly figure
fig = go.Figure()

# Add solid line for intervention accuracies
fig.add_trace(go.Scatter(
    x=[int(layer) for layer in layers],
    y=intervention_accuracies,
    mode="lines",
    line=dict(color="lightblue", width=3),
    name="Intervention Accuracy"
))

# Add dotted line for baseline accuracy
fig.add_trace(go.Scatter(
    x=[int(layer) for layer in layers],
    y=baseline_accuracies,
    mode="lines",
    line=dict(color="blue", width=3, dash="dash"),
    name="Baseline Accuracy"
))

# Update layout to match the style in the image
fig.update_layout(
    title="Capitalize",
    xaxis_title="Layer",
    yaxis_title="Accuracy",
    template="plotly_dark",
    font=dict(size=10),
    xaxis=dict(tickmode="linear", dtick=5),
    width=600,
    height=400
)

# Show the plot
fig.show()
```
