```python
with open(result_path + fs_file, "r") as f:
    data = json.load(f)

# Extract layers and after intervention accuracies
layers = list(data.keys())
layers = sorted(layers, key=int)  # Ensure layers are sorted numerically
intervention_accuracies = [data[layer]["intervention_topk"][0][1] for layer in layers]

# Extract baseline (before intervention) accuracies from layer 0
baseline_accuracy = data["0"]["clean_topk"][0][1]
baseline_accuracies = [baseline_accuracy] * len(layers)  # Same value across all layers

# Create Plotly figure
fig = go.Figure()

# Add solid line for intervention accuracies
fig.add_trace(go.Scatter(
    x=[int(layer) for layer in layers],
    y=intervention_accuracies,
    mode="lines",
    line=dict(color="lightblue", width=3),
    name="Intervention Accuracy"
))

# Add dotted line for baseline accuracy
fig.add_trace(go.Scatter(
    x=[int(layer) for layer in layers],
    y=baseline_accuracies,
    mode="lines",
    line=dict(color="blue", width=3, dash="dash"),
    name="Baseline Accuracy"
))

# Update layout to match the style in the image
fig.update_layout(
    title="Capitalize",
    xaxis_title="Layer",
    yaxis_title="Accuracy",
    template="plotly_dark",
    font=dict(size=10),
    xaxis=dict(tickmode="linear", dtick=5),
    width=600,
    height=400
)

# Show the plot
fig.show()
```

```python


# Load JSON data for both sources
file_path_1 = "fs_shuffled_results_layer_sweep.json"  # First JSON (Few Shots)
file_path_2 = "zs_results_layer_sweep.json"  # Second JSON (Zero Shot)

with open(result_path + file_path_1, "r") as f:
    data_1 = json.load(f)
with open(result_path + file_path_2, "r") as f:
    data_2 = json.load(f)

# Extract layers and accuracies for both JSON files
layers = sorted(data_1.keys(), key=int)  # Ensure layers are sorted numerically

intervention_accuracies_1 = [data_1[layer]["intervention_topk"][0][1] for layer in layers]
intervention_accuracies_2 = [data_2[layer]["intervention_topk"][0][1] for layer in layers]

# Extract baselines from layer 0
baseline_accuracy_1 = data_1["0"]["clean_topk"][0][1]
baseline_accuracy_2 = data_2["0"]["clean_topk"][0][1]

baseline_accuracies_1 = [baseline_accuracy_1] * len(layers)
baseline_accuracies_2 = [baseline_accuracy_2] * len(layers)

# Create subplots with specific titles
fig = make_subplots(
    rows=1, cols=2, subplot_titles=("Few Shots", "Zero Shot")
)

# Add solid line for intervention accuracies (Few Shots)
fig.add_trace(go.Scatter(
    x=[int(layer) for layer in layers],
    y=intervention_accuracies_1,
    mode="lines",
    line=dict(color="lightblue", width=3),
    name="Intervention Accuracy",
), row=1, col=1)

# Add dotted line for baseline accuracy (Few Shots)
fig.add_trace(go.Scatter(
    x=[int(layer) for layer in layers],
    y=baseline_accuracies_1,
    mode="lines",
    line=dict(color="blue", width=3, dash="dash"),
    name="Baseline Accuracy",
), row=1, col=1)

# Add solid line for intervention accuracies (Zero Shot)
fig.add_trace(go.Scatter(
    x=[int(layer) for layer in layers],
    y=intervention_accuracies_2,
    mode="lines",
    line=dict(color="lightblue", width=3),
    showlegend=False,
), row=1, col=2)

# Add dotted line for baseline accuracy (Zero Shot)
fig.add_trace(go.Scatter(
    x=[int(layer) for layer in layers],
    y=baseline_accuracies_2,
    mode="lines",
    line=dict(color="blue", width=3, dash="dash"),
    showlegend=False,
), row=1, col=2)

# Update layout for better spacing
fig.update_layout(
    title_text="Capitalize",
    xaxis_title="Layer",
    yaxis_title="Accuracy",
    template="plotly_dark",
    font=dict(size=10),
    width=1000,  # Increased width for side-by-side plots
    height=400,
    margin=dict(l=60, r=40, t=60, b=60)
)

# Show the plot
fig.show()
```
